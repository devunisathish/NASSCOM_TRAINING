### 8. Reverse a number and check whether it is a palindrome.

**Expected practice:** Arithmetic operations

Try solving this problem independently before running the example solution.

```python
number = int(input("Enter a non-negative integer: "))
if number < 0:
    print("Please enter a non-negative integer.")
else:
    original = number
    reversed_number = 0
    while number > 0:
        digit = number % 10
        reversed_number = reversed_number * 10 + digit
        number //= 10
    print("Reversed number:", reversed_number)
    print("Palindrome" if original == reversed_number else "Not a palindrome")
```

### 10. Print prime numbers within a given range.

**Expected practice:** Nested logic

Try solving this problem independently before running the example solution.

```python
start = int(input("Enter range start: "))
end = int(input("Enter range end: "))

primes = []
for number in range(max(2, start), end + 1):
    is_prime = True
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        primes.append(number)

print("Prime numbers:", primes)
```

### 11. Count vowels, consonants, digits, and spaces in a string.

**Expected practice:** String processing

Try solving this problem independently before running the example solution.

```python
text = input("Enter a string: ")
vowels = consonants = digits = spaces = 0

for char in text:
    if char.lower() in "aeiou":
        vowels += 1
    elif char.isalpha():
        consonants += 1
    elif char.isdigit():
        digits += 1
    elif char.isspace():
        spaces += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
```

### 17. Sort a list without using the built-in sort() method.

**Expected practice:** Algorithms

Try solving this problem independently before running the example solution.

```python
numbers = list(map(int, input("Enter integers separated by spaces: ").split()))

# Bubble sort
for i in range(len(numbers)):
    swapped = False
    for j in range(0, len(numbers) - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]
            swapped = True
    if not swapped:
        break

print("Sorted list:", numbers)
```

### 19. Find the most frequently occurring word in a sentence.

**Expected practice:** String processing

Try solving this problem independently before running the example solution.

```python
import re
from collections import Counter

sentence = input("Enter a sentence: ")
words = re.findall(r"\b[\w']+\b", sentence.lower())

if not words:
    print("No words found.")
else:
    counts = Counter(words)
    highest_count = max(counts.values())
    most_common = [word for word, count in counts.items() if count == highest_count]
    print("Most frequent word(s):", most_common)
    print("Frequency:", highest_count)
```

### 21. Create a calculator using separate functions for each operation.

**Expected practice:** Functions

Try solving this problem independently before running the example solution.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        raise ZeroDivisionError("Cannot divide by zero.")
    return a / b

a = float(input("First number: "))
operator = input("Operation (+, -, *, /): ").strip()
b = float(input("Second number: "))

operations = {"+": add, "-": subtract, "*": multiply, "/": divide}
try:
    if operator not in operations:
        print("Unsupported operation.")
    else:
        print("Result:", operations[operator](a, b))
except ZeroDivisionError as error:
    print(error)
```

### 22. Find the GCD and LCM using user-defined functions.

**Expected practice:** Functions, arithmetic

Try solving this problem independently before running the example solution.

```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return abs(a)

def lcm(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // gcd(a, b)

a = int(input("First integer: "))
b = int(input("Second integer: "))
print("GCD:", gcd(a, b))
print("LCM:", lcm(a, b))
```

### 23. Calculate factorial using recursion.

**Expected practice:** Recursion

Try solving this problem independently before running the example solution.

```python
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative integers.")
    if n in (0, 1):
        return 1
    return n * factorial(n - 1)

number = int(input("Enter a non-negative integer: "))
try:
    print(f"{number}! =", factorial(number))
except ValueError as error:
    print(error)
```

### 24. Implement linear search and binary search.

**Expected practice:** Algorithms

Try solving this problem independently before running the example solution.

```python
def linear_search(items, target):
    for index, item in enumerate(items):
        if item == target:
            return index
    return -1

def binary_search(sorted_items, target):
    low, high = 0, len(sorted_items) - 1
    while low <= high:
        mid = (low + high) // 2
        if sorted_items[mid] == target:
            return mid
        if sorted_items[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

numbers = list(map(int, input("Enter integers separated by spaces: ").split()))
target = int(input("Target value: "))
print("Linear search index:", linear_search(numbers, target))
sorted_numbers = sorted(numbers)
print("Sorted list for binary search:", sorted_numbers)
print("Binary search index in sorted list:", binary_search(sorted_numbers, target))
```

### 25. Implement bubble sort and selection sort.

**Expected practice:** Sorting

Try solving this problem independently before running the example solution.

```python
def bubble_sort(items):
    result = items.copy()
    for i in range(len(result)):
        for j in range(len(result) - i - 1):
            if result[j] > result[j + 1]:
                result[j], result[j + 1] = result[j + 1], result[j]
    return result

def selection_sort(items):
    result = items.copy()
    for i in range(len(result)):
        min_index = i
        for j in range(i + 1, len(result)):
            if result[j] < result[min_index]:
                min_index = j
        result[i], result[min_index] = result[min_index], result[i]
    return result

numbers = list(map(int, input("Enter integers separated by spaces: ").split()))
print("Bubble sort:", bubble_sort(numbers))
print("Selection sort:", selection_sort(numbers))
```

### 26. Create a student marks management system.

**Expected practice:** Lists, dictionaries

Try solving this problem independently before running the example solution.

```python
students = {}

def add_student():
    name = input("Student name: ").strip()
    marks = float(input("Marks (0–100): "))
    if not name:
        print("Name cannot be empty.")
    elif not 0 <= marks <= 100:
        print("Marks must be between 0 and 100.")
    else:
        students[name] = marks
        print("Student record saved.")

def view_students():
    if not students:
        print("No student records available.")
    else:
        for name, marks in students.items():
            print(f"{name}: {marks}")

while True:
    print("\n1. Add/update student  2. View all  3. Exit")
    choice = input("Choose: ").strip()
    if choice == "1":
        add_student()
    elif choice == "2":
        view_students()
    elif choice == "3":
        break
    else:
        print("Invalid choice.")
```

### 28. Store student records in a CSV file and retrieve them.

**Expected practice:** CSV processing

Try solving this problem independently before running the example solution.

```python
import csv
from pathlib import Path

file_name = "students.csv"
records = [
    {"name": "Asha", "roll_no": "101", "marks": "88"},
    {"name": "Rahul", "roll_no": "102", "marks": "79"},
]

with open(file_name, "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "roll_no", "marks"])
    writer.writeheader()
    writer.writerows(records)

print("Records saved to", Path(file_name).resolve())
with open(file_name, "r", newline="", encoding="utf-8") as file:
    for record in csv.DictReader(file):
        print(record)
```

### 30. Create a menu-driven library management program.

**Expected practice:** Functions, data structures

Try solving this problem independently before running the example solution.

```python
library = {}

def add_book():
    book_id = input("Book ID: ").strip()
    title = input("Book title: ").strip()
    if not book_id or not title:
        print("Book ID and title are required.")
    elif book_id in library:
        print("That book ID already exists.")
    else:
        library[book_id] = {"title": title, "available": True}
        print("Book added.")

def view_books():
    if not library:
        print("No books in the library.")
        return
    for book_id, book in library.items():
        status = "Available" if book["available"] else "Borrowed"
        print(f"{book_id}: {book['title']} — {status}")

def borrow_book():
    book_id = input("Book ID to borrow: ").strip()
    if book_id not in library:
        print("Book not found.")
    elif not library[book_id]["available"]:
        print("Book is already borrowed.")
    else:
        library[book_id]["available"] = False
        print("Book borrowed successfully.")

def return_book():
    book_id = input("Book ID to return: ").strip()
    if book_id not in library:
        print("Book not found.")
    elif library[book_id]["available"]:
        print("This book is already marked available.")
    else:
        library[book_id]["available"] = True
        print("Book returned successfully.")

while True:
    print("\n--- Library Menu ---")
    print("1. Add book  2. View books  3. Borrow book  4. Return book  5. Exit")
    choice = input("Choose an option: ").strip()
    if choice == "1":
        add_book()
    elif choice == "2":
        view_books()
    elif choice == "3":
        borrow_book()
    elif choice == "4":
        return_book()
    elif choice == "5":
        print("Goodbye!")
        break
    else:
        print("Invalid option. Choose 1–5.")
```
